```python
import nltk #importing the nltk module for nlp
from nltk.stem import WordNetLemmatizer #from stem module, we will import wordnetlemmatizer for lemmatization
lemma = WordNetLemmatizer() #assigning a variable for WordNetLemmatizer
print(lemma.lemmatize("better", pos="a")) #better word will be converted or brought into a adj
print(lemma.lemmatize("efforts"))
print(lemma.lemmatize("mice"))
print(lemma.lemmatize("feet"))
print(lemma.lemmatize("driving", pos="v")) #driving word will be brought into verb form
print(lemma.lemmatize("is", pos="v")) #is will be converted into verb
print(lemma.lemmatize("saw"))


```

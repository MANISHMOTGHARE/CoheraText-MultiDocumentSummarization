```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from google.colab import files

data = {
    "Model": ["PRIMERA", "BART-large", "PEGASUS", "T5-base", "Longformer", "TG-MultiSum"],
    "ROUGE-1": [33.81, 35.77, 36.55, 39.70, 32.46, 28.62],
    "ROUGE-2": [14.69, 14.72, 15.33, 23.53, 13.95, 11.10],
    "ROUGE-L": [24.29, 26.49, 26.21, 24.49, 24.95, 22.26]
}

df = pd.DataFrame(data)

df_melted = df.melt(id_vars="Model", var_name="Metric", value_name="Score")

sns.set(style="whitegrid")
palette = sns.color_palette("Set2")

#Bar Plot
plt.figure(figsize=(10, 6))
sns.barplot(x="Model", y="Score", hue="Metric", data=df_melted, palette=palette)
plt.title("ROUGE Scores Comparison (Bar Plot)", fontsize=14)
plt.xticks(rotation=15)
plt.ylabel("Score (%)")
plt.xlabel("Model")
plt.legend(title="ROUGE Metric")
plt.tight_layout()
plt.savefig("bar_plot.png")
files.download("bar_plot.png")
plt.show()



#Line Plot
plt.figure(figsize=(10, 6))
sns.lineplot(x="Model", y="Score", hue="Metric", data=df_melted, marker="o", palette=palette)
plt.title("ROUGE Scores Comparison (Line Plot)", fontsize=14)
plt.xticks(rotation=15)
plt.ylabel("Score (%)")
plt.xlabel("Model")
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig("ROUGE Scores Comparison (Line Plot).png")
files.download("ROUGE Scores Comparison (Line Plot).png")
plt.show()

```

```python
df = pd.read_csv(file_path, header=0)
df.columns = [
    "Timestamp",
    "Collector Name",
    "Newspaper",
    "Published Date",
    "Article URL",
    "Headline",
    "Article Content",
    "Human Summary",
    "Category"
]

print(f"Initial rows: {len(df)}")
df.head()
```

```python
# Step 4: Basic Cleaning
from dateutil import parser

# Drop empty rows
df = df.dropna(subset=["Article URL", "Article Content", "Headline"])

# Convert dates
df["Published Date"] = pd.to_datetime(
    df["Published Date"],
    errors="coerce",
    format="mixed"
)

# Remove invalid dates
df = df[df["Published Date"].notna()]
print(f"Rows after cleaning: {len(df)}")
```

```python
# Step 5: Deduplication
# Remove exact duplicates
df = df.drop_duplicates(subset=["Article URL", "Article Content"])

# Near-duplicates using first 200 characters
from simhash import Simhash

def get_simhash(text):
    return Simhash(str(text)[:200].split()).value

df["simhash"] = df["Article Content"].apply(get_simhash)
df = df.drop_duplicates(subset=["simhash"])
df = df.drop(columns=["simhash"])

print(f"Rows after deduplication: {len(df)}")
```

```python
# Step 6: Text Cleaning
from bs4 import BeautifulSoup
import re

def clean_content(text):
    # Basic cleaning
    text = BeautifulSoup(str(text), "html.parser").get_text()
    text = re.sub(r'http\S+', '', text)  # Remove URLs
    text = text.encode("ascii", "ignore").decode()
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df["Article Content"] = df["Article Content"].apply(clean_content)
df["Headline"] = df["Headline"].apply(clean_content)

# Standardize Newspaper Names
newspaper_mapping = {
    "the hindu": "The Hindu",
    "The Sangai Express": "The Sangai Express",
    # Add more mappings as needed
}
df["Newspaper"] = df["Newspaper"].str.strip().replace(newspaper_mapping)

# Clean Categories
category_mapping = {
    "Science and Technology": "Technology",
    "National News": "National",
    # Add more mappings
}
df["Category"] = df["Category"].str.strip().replace(category_mapping)
```

```python
from sklearn.model_selection import train_test_split
import numpy as np

# --------------------------------------------------
# Step 1: Filter categories with sufficient samples
# --------------------------------------------------
MIN_SAMPLES_PER_CLASS = 10  # Minimum 10 samples per category
category_counts = df["Category"].value_counts()

# Keep only categories meeting threshold
valid_categories = category_counts[category_counts >= MIN_SAMPLES_PER_CLASS].index
df_filtered = df[df["Category"].isin(valid_categories)].copy()

print("Categories after filtering:", df_filtered["Category"].nunique())

# --------------------------------------------------
# Step 2: Merge residual rare categories into "Other"
# --------------------------------------------------
# Calculate post-filtering counts
post_filter_counts = df_filtered["Category"].value_counts()

# Merge categories with <10 samples into "Other"
df_filtered["Category"] = np.where(
    df_filtered["Category"].map(post_filter_counts) < MIN_SAMPLES_PER_CLASS,
    "Other",
    df_filtered["Category"]
)

# Final category validation
final_counts = df_filtered["Category"].value_counts()
print("\nFinal category distribution:\n", final_counts)

# --------------------------------------------------
# Step 3: Stratified split with safety checks
# --------------------------------------------------
def safe_stratified_split(data, test_size):
    """Handles edge cases where stratification might fail"""
    try:
        return train_test_split(
            data,
            test_size=test_size,
            stratify=data["Category"],
            random_state=42
        )
    except ValueError:
        # Fallback to non-stratified split if stratification fails
        return train_test_split(
            data,
            test_size=test_size,
            random_state=42
        )

# First split: Train (80%) vs Temp (20%)
train_df, temp_df = safe_stratified_split(df_filtered, 0.2)

# Second split: Val (10%) vs Test (10%)
val_df, test_df = safe_stratified_split(temp_df, 0.5)

print("\nFinal split counts:")
print(f"Train: {len(train_df)} | Val: {len(val_df)} | Test: {len(test_df)}")
print("\nValidation set category counts:\n", val_df["Category"].value_counts())
```

```python
from transformers import BartTokenizer, BartForConditionalGeneration
from datasets import Dataset
import pandas as pd

# 1. Load Tokenizer
tokenizer = BartTokenizer.from_pretrained("facebook/bart-large-cnn")

# 2. Preprocessing Function
def preprocess_function(examples):
    inputs = tokenizer(
        examples["Article Content"],
        max_length=1024,
        truncation=True,
        padding="max_length"
    )
    labels = tokenizer(
        examples["Human Summary"],
        max_length=256,
        truncation=True,
        padding="max_length"
    )
    return {
        "input_ids": inputs["input_ids"],
        "attention_mask": inputs["attention_mask"],
        "labels": labels["input_ids"]
    }

# 3. Prepare Dataset
train_dataset = Dataset.from_pandas(train_df)
train_dataset = train_dataset.map(
    preprocess_function,
    batched=True,
    remove_columns=["Article Content", "Human Summary", "__index_level_0__"]  # Remove raw text columns
)

# 4. Update Training Arguments
training_args = Seq2SeqTrainingArguments(
    output_dir="./results",
    per_device_train_batch_size=2,
    remove_unused_columns=False,  # Critical fix
    # Keep other arguments...
)

# 5. Reload Model
model = BartForConditionalGeneration.from_pretrained("facebook/bart-large-cnn")

# 6. Reinitialize Trainer
trainer = WeightedTrainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=Dataset.from_pandas(val_df).map(preprocess_function, batched=True)
)

# Now run trainer.train()
```

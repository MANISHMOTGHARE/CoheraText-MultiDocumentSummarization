```python
import torch
from torch.utils.data import Dataset, DataLoader, random_split
from transformers import BartForConditionalGeneration, PegasusForConditionalGeneration, T5ForConditionalGeneration
from transformers import LongformerForSequenceClassification, BigBirdForSequenceClassification, AdamW, BartTokenizer
from tqdm import tqdm
import pandas as pd
from rouge_score import rouge_scorer
import gc
from torch.optim import AdamW as torch_AdamW
import torch
from torch.utils.data import Dataset, DataLoader

# ... rest of the code remains the same ...

# Ensure you're using GPU if available
device = "cuda" if torch.cuda.is_available() else "cpu"

# Mixed Precision Training
scaler = torch.cuda.amp.GradScaler()

# Load your tokenizer and define custom dataset class
tokenizer = BartTokenizer.from_pretrained('facebook/bart-base')
max_length = 512

class CustomDataset(Dataset):
    def __init__(self, data, tokenizer, max_length):
        self.data = data
        self.tokenizer = tokenizer
        self.max_length = max_length

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        input_text = self.data.iloc[idx]['article']
        target_text = self.data.iloc[idx]['highlights']

        input_tokens = self.tokenizer(input_text, max_length=self.max_length, padding='max_length', truncation=True, return_tensors="pt")
        target_tokens = self.tokenizer(target_text, max_length=self.max_length, padding='max_length', truncation=True, return_tensors="pt")

        return {
            'input_ids': input_tokens['input_ids'].squeeze(),
            'attention_mask': input_tokens['attention_mask'].squeeze(),
            'labels': target_tokens['input_ids'].squeeze()
        }

# Load your data and limit it to 10,000 for training, 8,000 for validation, and 2,000 for testing
train_data = pd.read_csv('/Users/taaran/Jupyter Notebooks/untitled folder/train.csv').sample(6000)
val_data = pd.read_csv('/Users/taaran/Jupyter Notebooks/untitled folder/validation.csv').sample(3000)
test_data = pd.read_csv('/Users/taaran/Jupyter Notebooks/untitled folder/test.csv').sample(1000)

# Create Dataset instances
train_dataset = CustomDataset(train_data, tokenizer, max_length)
val_dataset = CustomDataset(val_data, tokenizer, max_length)
test_dataset = CustomDataset(test_data, tokenizer, max_length)

# Reduce the batch size to prevent memory overflow
train_loader = DataLoader(train_dataset, batch_size=4, shuffle=True)  # Batch size reduced to 4
val_loader = DataLoader(val_dataset, batch_size=2)
test_loader = DataLoader(test_dataset, batch_size=2)

# Define function for training
def train_model(model, train_loader, optimizer, criterion):
    model.train()
    total_loss = 0
    for batch in tqdm(train_loader):
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        labels = batch['labels'].to(device)

        optimizer.zero_grad()
        
        # Mixed Precision Training
        with torch.cuda.amp.autocast():
            outputs = model(input_ids=input_ids, attention_mask=attention_mask, labels=labels)
            loss = outputs.loss

        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()

        total_loss += loss.item()

    return total_loss / len(train_loader)

# Define evaluation function
def evaluate_model(model, eval_loader):
    model.eval()
    predictions, references = [], []
    with torch.no_grad():
        for batch in eval_loader:
            input_ids = batch['input_ids'].to(device)
            attention_mask = batch['attention_mask'].to(device)
            labels = batch['labels'].to(device)

            outputs = model.generate(input_ids=input_ids, attention_mask=attention_mask, max_length=max_length)
            preds = tokenizer.batch_decode(outputs, skip_special_tokens=True)
            refs = tokenizer.batch_decode(labels, skip_special_tokens=True)

            predictions.extend(preds)
            references.extend(refs)
    
    # Calculate ROUGE scores
    rouge_score = rouge_scorer.compute(predictions=predictions, references=references, rouge_types=['rouge1', 'rougeL'])
    return rouge_score

# Free up memory after each epoch
def free_memory():
    torch.cuda.empty_cache()
    gc.collect()

# Initialize models (pre-trained models)
models = {
    'BART': BartForConditionalGeneration.from_pretrained('facebook/bart-base').to(device),
    'Pegasus': PegasusForConditionalGeneration.from_pretrained('google/pegasus-large').to(device),
    'T5': T5ForConditionalGeneration.from_pretrained('t5-base').to(device),
    'Longformer': LongformerForSequenceClassification.from_pretrained('allenai/longformer-base-4096').to(device),
    'BigBird': BigBirdForSequenceClassification.from_pretrained('google/bigbird-roberta-base').to(device)
}

# Define optimizer and loss criterion
criterion = torch.nn.CrossEntropyLoss(ignore_index=tokenizer.pad_token_id)

# Dictionary to store results
results = []

# Training and evaluation loop for each model
for model_name, model in models.items():
    print(f"\nTraining {model_name}...")
    optimizer = AdamW(model.parameters(), lr=3e-5)

    # Train for 3 epochs (adjustable)
    for epoch in range(1):
        print(f"Epoch {epoch+1} for {model_name}")
        train_loss = train_model(model, train_loader, optimizer, criterion)
        print(f"{model_name} - Training Loss: {train_loss}")

        # Evaluate the model on validation set
        rouge_scores_val = evaluate_model(model, val_loader)
        print(f"{model_name} - Validation ROUGE Scores: {rouge_scores_val}")

    # Evaluate on test set after training
    rouge_scores_test = evaluate_model(model, test_loader)
    print(f"{model_name} - Test ROUGE Scores: {rouge_scores_test}")

    # Store the results (Validation and Test Scores)
    results.append({
        'Model': model_name,
        'ROUGE-1 (Validation)': rouge_scores_val['rouge1'].mid.fmeasure,
        'ROUGE-2 (Validation)': rouge_scores_val['rouge2'].mid.fmeasure,
        'ROUGE-L (Validation)': rouge_scores_val['rougeL'].mid.fmeasure,
        'ROUGE-1 (Test)': rouge_scores_test['rouge1'].mid.fmeasure,
        'ROUGE-2 (Test)': rouge_scores_test['rouge2'].mid.fmeasure,
        'ROUGE-L (Test)': rouge_scores_test['rougeL'].mid.fmeasure
    })
    
    # Clear cache after training each model
    free_memory()

# Convert results to a DataFrame
results_df = pd.DataFrame(results)

# Display the evaluation matrix in tabular form
print("\nEvaluation Matrix:")
print(results_df)
```

```python
# Import the necessary libraries
import pandas as pd
from transformers import PegasusTokenizer, PegasusForConditionalGeneration
from rouge_score import rouge_scorer

# Load the dataset
file_path = '/content/synthetic_news_dataset.csv'
df = pd.read_csv(file_path)

# Load the Pegasus model and tokenizer
model_name = 'google/pegasus-xsum'
tokenizer = PegasusTokenizer.from_pretrained(model_name)
model = PegasusForConditionalGeneration.from_pretrained(model_name)

# Function to generate summaries
def generate_summary(text):
    inputs = tokenizer.encode("summarize: " + text, return_tensors='pt', max_length=512, truncation=True)
    summary_ids = model.generate(inputs, max_length=60, min_length=10, length_penalty=2.0, num_beams=4, early_stopping=True)
    summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)
    return summary
```

```python

# Apply the function to the 'Content in Detail of News Article' column and create a new column 'Generated Summary'
df['Generated Summary'] = df['News Content in Detail'].apply(generate_summary)

# Save the updated dataframe to a new CSV file
output_file_path = '/content/news_records.csv'
df.to_csv(output_file_path, index=False)
```

```python

# Initialize ROUGE scorer
scorer = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)

# Function to calculate ROUGE scores
def calculate_rouge_scores(ref, pred):
    scores = scorer.score(ref, pred)
    return scores['rouge1'].fmeasure, scores['rouge2'].fmeasure, scores['rougeL'].fmeasure

# Calculate ROUGE scores for each summary
rouge_scores = df.apply(lambda row: calculate_rouge_scores(row['News Content Summary (20 words)'], row['Generated Summary']), axis=1)

# Create separate columns for each ROUGE score
df['ROUGE-1'] = rouge_scores.apply(lambda x: x[0])
df['ROUGE-2'] = rouge_scores.apply(lambda x: x[1])
df['ROUGE-L'] = rouge_scores.apply(lambda x: x[2])

# Save the updated dataframe with ROUGE scores to a new CSV file
output_file_path_with_rouge = '/content/news_records.csv'
df.to_csv(output_file_path_with_rouge, index=False)

print(f"Summaries generated and ROUGE scores calculated. Data saved to {output_file_path_with_rouge}")
```

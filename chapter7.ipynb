```python
import nltk
from nltk.tokenize import word_tokenize #for divide the sentence into words
from nltk.tag import pos_tag #position tag is being imported for assisgning a position for each word
a= "Narendra Modi on Tuesday completed his world tour to countries like USA, London, Canada, Russia, China." #Sample text

words= word_tokenize(a) #converting the sentences into word type
postags=pos_tag(words) #adding position tag to each word which is extracted
NER = nltk.ne_chunk(postags,binary=False) #apply the NER condition to the sample text and storing it into a new variable
print(NER)
```

```python
from transformers import PegasusTokenizer, PegasusForConditionalGeneration
from datasets import load_dataset
from rouge_score import rouge_scorer

# Load the Pegasus model and tokenizer
model_name = "google/pegasus-xsum"
tokenizer = PegasusTokenizer.from_pretrained(model_name)
model = PegasusForConditionalGeneration.from_pretrained(model_name)

# Load the XSum dataset
dataset = load_dataset("xsum", split="test")

# Sample a text from the dataset
sample_text = dataset[0]['document']

# Tokenize and summarize the text
inputs = tokenizer(sample_text, truncation=True, padding="longest", return_tensors="pt")
summary_ids = model.generate(inputs["input_ids"])
summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)

print("Original Document: ", sample_text)
print("Generated Summary: ", summary)
```

```python
# Function to calculate ROUGE scores
def calculate_rouge(reference_summary, generated_summary):
    scorer = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)
    scores = scorer.score(reference_summary, generated_summary)
    return scores

# Get the reference summary from the dataset
reference_summary = dataset[0]['summary']

# Calculate ROUGE scores
rouge_scores = calculate_rouge(reference_summary, summary)
```

```python
# Output ROUGE scores as percentages
print(f"ROUGE-1: {rouge_scores['rouge1'].fmeasure * 100:.2f}%")
print(f"ROUGE-2: {rouge_scores['rouge2'].fmeasure * 100:.2f}%")
print(f"ROUGE-L: {rouge_scores['rougeL'].fmeasure * 100:.2f}%")
```

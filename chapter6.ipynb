```python
import nltk
from nltk import pos_tag #position tag from POS
from nltk import word_tokenize#divide the sentence into words
text='''To remove the chunk from the extracted chunk is called as chink.
     The pattern of chink is equivalent to that of chunking but the syntax of them very much varies.''' #sample text
tokenized = word_tokenize(text) #converting the sentences into word type
tagged_words = pos_tag(tokenized) #adding position tag to each word which is extracted

def ab(): #defining a function which can be used again and again
    try: # try and except if at all any errors may occur
        for i in tokenized[5:]: #parsing through the chunked sentences
            words = nltk.word_tokenize(i)
            tagged = nltk.pos_tag(words)
            chunkGram = r"""Chunk: {<.*>+} 
                                    }<VB.?|IN|DT|TO>+{""" #adding the chink condition to chunk
            chunkParser = nltk.RegexpParser(chunkGram)
            chunked = chunkParser.parse(tagged)
            for subtree in chunked.subtrees(): #parsing through the chunked sentences
               if subtree.label() == 'Chunk': #checking the proper condition for the Chunk 
                 print(subtree) #printing the optimized chink
    except Exception as e:
        print(str(e))
ab()
```

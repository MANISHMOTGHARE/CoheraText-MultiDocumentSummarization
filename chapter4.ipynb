```python
import nltk #importing the nltk for tokenization
from nltk.tokenize import sent_tokenize, word_tokenize #importing word and sentence tokenization from nltk library
text='''Part of speech are those type of words which are used to 
compose a sentence into a logical sentence 
and give the correct meaning of the sentence.
Even though part of speech consists of different parts,
it still remains to be complicate to understand.''' #text is a variable which consist of sample sentence
tokenized = sent_tokenize(text) #initializing the text into sentence tokenization and assigning it into new variable
def a(): #defining an function named a which can be used again and again
    try: #to check if the tokenization of working properly or if there are any errors.
        for i in tokenized: #setting a loop so that one word after another will be checked aand find its part of speech
            words = nltk.word_tokenize(i) #break the sentences into words
            tagged = nltk.pos_tag(words) #adding the parts of speech to each word
            print(tagged) #printing the parts of speech

    except Exception as e: #if any errors, expection will throw the cause
        print(e) #prints the expection
a() #calling the function
```

```python
import gspread
from google.colab import auth
from google.auth import default

# Authenticate
auth.authenticate_user()
creds, _ = default()
gc = gspread.authorize(creds)

# Open the Google Sheets
spreadsheet_url = "https://docs.google.com/spreadsheets/d/1GGX3plNkunqRFSxbwUqpV4yJQGDn8gA8BGqFFovPcrc/edit?gid=0"
spreadsheet = gc.open_by_url(spreadsheet_url)
worksheet = spreadsheet.get_worksheet(0)  # Open the first sheet
```

```python
# Get the headers from the first row
headers = worksheet.row_values(1)  # First row (header)
print("Headers:", headers)
print(f"Number of columns: {len(headers)}, Unique columns: {len(set(headers))}")
```

```python
# Get all values from the Google Sheet
all_cells = worksheet.get_all_values()  # This reads all rows from the sheet
```

```python
import pandas as pd

# Convert the sheet data to a DataFrame
data = pd.DataFrame(all_cells[1:], columns=all_cells[0])  # Skip the first row (header)

# Display the first 5 rows to confirm it loaded correctly
print(data.head())
```

```python
# Print the first few rows of the entire sheet to inspect
for row in all_cells[:5]:
    print(row)
```

```python
print(data.columns)
```

```python
# Use the second row as the header (actual column names)
data = pd.DataFrame(all_cells[2:], columns=all_cells[1])  # Skip the first row, use the second row as headers

# Display the first few rows to confirm
print(data.head())
print(data.columns)
```

```python
print(data.columns)
```

```python
description = all_cells[0][0]  # Store the first row (description) separately
print("Description:", description)
```

```python
from transformers import PegasusTokenizer, PegasusForConditionalGeneration

# Load PEGASUS model
model_name = "google/pegasus-xsum"
tokenizer = PegasusTokenizer.from_pretrained(model_name)
model = PegasusForConditionalGeneration.from_pretrained(model_name)

# Summarize each content
summaries = []
for text in data['Content']:
    if len(text) > 50:  # Only summarize long content
        inputs = tokenizer(text, return_tensors="pt", max_length=512, truncation=True)
        summary_ids = model.generate(inputs.input_ids, max_length=60, num_beams=5, early_stopping=True)
        summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)
        summaries.append(summary)
    else:
        summaries.append(text)

# Add summaries to DataFrame
data['Generated Summary'] = summaries
print(data[['Headline', 'Generated Summary']].head())
```

```python
from rouge_score import rouge_scorer
import pandas as pd
```

```python
scorer = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)
```

```python
# Create lists to store scores
rouge1_scores, rouge2_scores, rougeL_scores = [], [], []

# Iterate through each row in the DataFrame
for i, row in data.iterrows():
    generated_summary = row['Generated Summary']
    human_summary = row['Human Summary']  # Replace with your human summary column

    # Compute ROUGE scores
    scores = scorer.score(human_summary, generated_summary)
    rouge1_scores.append(scores['rouge1'].fmeasure)
    rouge2_scores.append(scores['rouge2'].fmeasure)
    rougeL_scores.append(scores['rougeL'].fmeasure)

# Add ROUGE scores to the DataFrame
data['ROUGE-1'] = rouge1_scores
data['ROUGE-2'] = rouge2_scores
data['ROUGE-L'] = rougeL_scores

# Display the first few rows with ROUGE scores
print(data[['Headline', 'Generated Summary', 'Human Summary', 'ROUGE-1', 'ROUGE-2', 'ROUGE-L']].head())
```

```python
average_rouge1 = sum(rouge1_scores) / len(rouge1_scores)
average_rouge2 = sum(rouge2_scores) / len(rouge2_scores)
average_rougeL = sum(rougeL_scores) / len(rougeL_scores)

print(f"Average ROUGE-1 Score: {average_rouge1:.4f}")
print(f"Average ROUGE-2 Score: {average_rouge2:.4f}")
print(f"Average ROUGE-L Score: {average_rougeL:.4f}")
```

```python
import matplotlib.pyplot as plt

# Histogram of ROUGE-1 scores
plt.hist(rouge1_scores, bins=10, alpha=0.7, label='ROUGE-1')
plt.hist(rouge2_scores, bins=10, alpha=0.7, label='ROUGE-2')
plt.hist(rougeL_scores, bins=10, alpha=0.7, label='ROUGE-L')
plt.title("ROUGE Score Distribution")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```
